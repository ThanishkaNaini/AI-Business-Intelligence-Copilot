```python
schema = """
Table: sales

Columns:
row_id INTEGER
order_id TEXT
order_date TIMESTAMP
ship_date TIMESTAMP
ship_mode TEXT
customer_id TEXT
customer_name TEXT
segment TEXT
country_region TEXT
city TEXT
state_province TEXT
postal_code TEXT
region TEXT
product_id TEXT
category TEXT
sub_category TEXT
product_name TEXT
sales REAL
quantity INTEGER
discount REAL
profit REAL
"""
```

```python
system_prompt = f"""
You are a Business Intelligence SQL assistant.

Your job is to convert business questions into SQLite SQL queries.

Use only this database schema:

{schema}

Rules:
1. Use only the sales table.
2. Generate valid SQLite SQL.
3. Do not invent columns.
4. Return only the SQL query.
5. Use SUM(sales) for total sales.
6. Use SUM(profit) for total profit.
7. Use GROUP BY when comparing regions, categories, customers, products, or segments.
"""
```

```python
def generate_sql(question):
    response = client.responses.create(
        model="gpt-5.6-luna",
        instructions=system_prompt,
        input=question
    )

    return response.output_text
```

```python
generate_sql("Which region has the highest sales?")
```

```python
def execute_sql(sql):
    sql_clean = sql.strip().lower()

    if not sql_clean.startswith("select"):
        raise ValueError("Only SELECT queries are allowed.")

    return pd.read_sql_query(sql, conn)
```

```python
def explain_result(question, result):
    result_text = result.to_string(index=False)

    response = client.responses.create(
        model="gpt-5.6-luna",
        input=f"""
You are a Business Intelligence analyst.

Business question:
{question}

Query result:
{result_text}

Explain the result in 2-3 clear business sentences.
Use only the numbers shown in the query result.
Do not invent any additional facts.
"""
    )

    return response.output_text
```

```python
question = "Which category has the highest profit?"

sql = generate_sql(question)
result = execute_sql(sql)

explanation = explain_result(question, result)

print(explanation)
```

```python
def business_intelligence_copilot(question):
    try:
        sql = generate_sql(question)
        result = execute_sql(sql)
        explanation = explain_result(question, result)

        print("QUESTION:")
        print(question)

        print("\nGENERATED SQL:")
        print(sql)

        print("\nRESULT:")
        display(result)

        print("\nBUSINESS EXPLANATION:")
        print(explanation)

    except Exception as e:
        print("Sorry, I could not answer that question.")
        print("Error:", e)
```

```python
def generate_executive_summary():
    summary_query = """
    SELECT
        SUM(sales) AS total_sales,
        SUM(profit) AS total_profit,
        (SUM(profit) / SUM(sales)) * 100 AS profit_margin_percent,
        COUNT(DISTINCT order_id) AS total_orders,
        COUNT(DISTINCT customer_id) AS total_customers
    FROM sales;
    """

    summary_data = execute_sql(summary_query)

    return summary_data
```

```python
generate_executive_summary()
```

```python
def ai_executive_summary():
    kpis = generate_executive_summary()

    kpi_text = kpis.to_string(index=False)

    response = client.responses.create(
        model="gpt-5.6-luna",
        input=f"""
You are a Business Intelligence analyst.

Here are the company's key performance indicators:

{kpi_text}

Write a concise executive summary in 3-4 sentences.

Rules:
1. Mention total sales, total profit, profit margin, total orders, and total customers.
2. Use clear business language.
3. Do not invent any information.
4. Only use the data provided.
"""
    )

    return response.output_text
```

```python
ai_executive_summary()
```

```python
def get_region_summary():
    query = """
    SELECT
        region,
        SUM(sales) AS total_sales,
        SUM(profit) AS total_profit,
        (SUM(profit) / SUM(sales)) * 100 AS profit_margin_percent
    FROM sales
    GROUP BY region
    ORDER BY total_sales DESC;
    """

    return execute_sql(query)
```

```python
get_region_summary()
```

```python
def get_category_summary():
    query = """
    SELECT
        category,
        SUM(sales) AS total_sales,
        SUM(profit) AS total_profit,
        (SUM(profit) / SUM(sales)) * 100 AS profit_margin_percent
    FROM sales
    GROUP BY category
    ORDER BY total_sales DESC;
    """

    return execute_sql(query)
```

```python
get_category_summary()
```

```python
def advanced_executive_summary():
    kpis = generate_executive_summary()
    regions = get_region_summary()
    categories = get_category_summary()

    response = client.responses.create(
        model="gpt-5.6-luna",
        input=f"""
You are a Business Intelligence analyst.

Overall KPIs:
{kpis.to_string(index=False)}

Regional Performance:
{regions.to_string(index=False)}

Category Performance:
{categories.to_string(index=False)}

Write a concise executive summary in 5-6 sentences.

Include:
1. Overall sales, profit, and profit margin.
2. Strongest region.
3. Weakest region based on profit margin.
4. Strongest product category.
5. Weakest product category based on profit margin.
6. One clear business takeaway.

Do not invent any data.
Use only the information provided.
"""
    )

    return response.output_text
```

```python
advanced_executive_summary()
```

```python
def get_monthly_summary():
    query = """
    SELECT
        strftime('%Y-%m', order_date) AS month,
        SUM(sales) AS total_sales,
        SUM(profit) AS total_profit
    FROM sales
    GROUP BY month
    ORDER BY month;
    """

    return execute_sql(query)
```

```python
get_monthly_summary()
```

```python
def get_yearly_summary():
    query = """
    SELECT
        strftime('%Y', order_date) AS year,
        SUM(sales) AS total_sales,
        SUM(profit) AS total_profit
    FROM sales
    GROUP BY year
    ORDER BY year;
    """

    return execute_sql(query)
```

```python
get_yearly_summary()
```

```python
yearly_summary = get_yearly_summary()

yearly_summary["sales_growth_percent"] = (
    yearly_summary["total_sales"].pct_change() * 100
)

yearly_summary["profit_growth_percent"] = (
    yearly_summary["total_profit"].pct_change() * 100
)

yearly_summary
```

```python
def get_yearly_growth():
    yearly = get_yearly_summary()

    yearly["sales_growth_percent"] = (
        yearly["total_sales"].pct_change() * 100
    )

    yearly["profit_growth_percent"] = (
        yearly["total_profit"].pct_change() * 100
    )

    return yearly
```

```python
get_yearly_growth()
```

```python
def final_executive_summary():
    kpis = generate_executive_summary()
    regions = get_region_summary()
    categories = get_category_summary()
    yearly_growth = get_yearly_growth()

    response = client.responses.create(
        model="gpt-5.6-luna",
        input=f"""
You are a Business Intelligence analyst.

Overall KPIs:
{kpis.to_string(index=False)}

Regional Performance:
{regions.to_string(index=False)}

Category Performance:
{categories.to_string(index=False)}

Yearly Performance and Growth:
{yearly_growth.to_string(index=False)}

Write a professional executive summary in 6-8 sentences.

Include:
1. Overall sales, profit, and profit margin.
2. Latest-year sales growth.
3. Latest-year profit growth.
4. Strongest region.
5. Weakest region based on profit margin.
6. Strongest product category.
7. Weakest product category based on profit margin.
8. One clear business recommendation.

Do not invent data.
Use only the information provided.
"""
    )

    return response.output_text
```
